## 🔹 1. Lineáris regressziós dinamikabecslés hatása a "beszűkülésre"

### 📌 Emlékeztető:

A rendszered minden lépésben újrabecsli a dinamikát a következő alakban:

$$
x[k+1] = A x[k] + B u[k]
$$

Ahol:

* Az $A$ és $B$ mátrixokat **lineáris regresszióval** becsülöd újra minden időlépésben.
* A regresszió alapját az összes eddigi adat képezi: $\{x[0], u[0], x[1]\}, \{x[1], u[1], x[2]\}, \dots$

---

### 🔍 Mi történik, ha a rendszer sokáig „jó” állapotban marad?

Ha a szabályozó hatékonyan tartja a rendszert egy „ideális” állapot körül, akkor:

* Az $x[k]$, $u[k]$ és $x[k+1]$ pontok **csak kis mértékben változnak** az időben.
* Ez azt jelenti, hogy a regresszió egy **lokálisan nagyon szűk tartományban** gyűjt adatot.

Ennek következményei:

---

### ⚠️ **A regresszió torzulása: beszűkült látótér**

1. **Szimmetrikus adatfedés megszűnése**:

   * A regresszió egy sokdimenziós téren belül dolgozik.
   * Ha az adatok egy kis környezetre koncentrálódnak, a becsült $A, B$ mátrixok **csak ott jók** — más állapotokban pontatlanok lehetnek.

2. **Illesztési probléma degenerálódása**:

   * Kevés és hasonló adat esetén az $X$-mátrix (design matrix) **rosszul kondicionált** lehet.
   * Ez numerikus instabilitást vagy túltanulást okoz.

3. **Generalizációs hiba növekedése**:

   * A modell „megtanulja”, hogy a világ nagyon egyszerű és csak egyféle viselkedést mutat.
   * Amint új állapotba kerül a rendszer, a becsült dinamika **nem írja le pontosan a viselkedést**.

---

### 📘 Egy intuitív példa:

Képzeld el, hogy a valós rendszered valójában **nem lineáris**, például:

$$
x[k+1] = A x[k] + B u[k] + \epsilon \sin(x[k])
$$

Ha a szabályozó mindig olyan $x[k]$ környékén van, ahol $\sin(x[k]) \approx 0$, akkor a becslés **sosem „látja”** a szinusz tag hatását. A regresszió így azt „tanulja meg”, hogy a rendszer *tényleg* lineáris. Ez a beszűkülés klasszikus esete.

---

### 🔧 Analitikus nézőpont: feltételes kovarianciamátrix

A lineáris regresszió becslése (pl. legkisebb négyzetek):

$$
\begin{bmatrix} A & B \end{bmatrix} = Y X^\dagger
$$

Ahol:

* $X = \begin{bmatrix} x[0] & x[1] & \cdots \end{bmatrix}$
* $Y = \begin{bmatrix} x[1] & x[2] & \cdots \end{bmatrix}$
* $X^\dagger$: Moore–Penrose pszeudo-inverz

Ha az $X$ mátrix soraiban nagyon kicsi a változatosság (a variancia minimális), akkor:

* A kovariancia mátrix $X X^T$ **nem invertálható vagy gyenge rangú**,
* azaz a becsült paraméterek nagyon **bizonytalanok vagy torzítottak** lesznek.

## 🔍 Elméleti alap: dinamikusan változó regressziós célfüggvény

Az OLS-becslés minden lépésben újrailleszti:

$$
\hat{A}_k, \hat{B}_k = \arg\min_{A, B} \sum_{i=0}^k \left\| x[i+1] - A x[i] - B u[i] \right\|^2
$$

A döntéshozás pedig az aktuális $\hat{A}_k, \hat{B}_k$ alapján történik, pl. MPC vagy policy network révén. Ez azt jelenti, hogy a **célfüggvény maga is időben változik**, méghozzá **endogén módon**: a becslés **maga befolyásolja** a jövőbeli mintákat, amikből tanulni fog.

---

## 📈 Ezen dinamika következményei:

### 1. **Önreferenciális torzítás** (self-induced bias)

* Mivel az $A, B$ becslését a szabályozó döntései által generált adatból tanuljuk,
* a rendszer olyan adatot „gyárt”, amely **jó a jelenlegi becslés szempontjából**,
* ez viszont **nem garantálja**, hogy jó lesz általánosításhoz vagy robusztus működéshez.

### 2. **Exploráció és ekszploitatív viselkedés feszültsége**

* Ha a rendszer túl gyorsan stabilizálódik egy „jó” állapot körül, a tanulás **csak lokálisan történik**,
* ez csökkenti a becslés általános érvényességét: az $A, B$ mátrixok **nem ismerik a globális dinamikát**,
* másrészt ha túl sokat „explorálunk”, akkor romlik a szabályozás teljesítménye — tehát van egy trade-off.

### 3. **Tanulás és irányítás közti visszacsatolt egyensúly**

* A tanulás célja: *jó becslést adni*,
* A szabályozás célja: *jó pályán tartani a rendszert*,
* De ezek **egymásra hatnak**: a jó szabályozás „rossz” adatokat adhat a tanuláshoz (pl. túl kevés variancia).

## 🧠 Hogyan lehet ezt *analitikusan* modellezni?

Egy lehetséges modell:

1. Tegyük fel, hogy az igaz rendszer:

$$
x[k+1] = A_{\text{true}} x[k] + B_{\text{true}} u[k] + w[k]
$$

2. De mi csak:

$$
x[k+1] \approx \hat{A}_k x[k] + \hat{B}_k u[k]
$$

3. A döntés:

$$
u[k] = \pi(x[k]; \hat{A}_k, \hat{B}_k)
$$

Ez egy olyan **dinamikus rendszer**, ahol az állapot: $(x[k], \hat{A}_k, \hat{B}_k)$, és a frissítés:

* $x[k+1] = A_{\text{true}} x[k] + B_{\text{true}} \pi(x[k]; \hat{A}_k, \hat{B}_k) + w[k]$
* $(\hat{A}_{k+1}, \hat{B}_{k+1}) = \text{OLS}(\{x[0:k+1], u[0:k]\})$

Ez egy **hibrid rendszer**, amelyben a becslés és az állapot frissítése **összefonódik**.

## 🧭 Most induljunk el az első kérdés mentén

### ❓ Alapkérdés:

**„Mi garantálja, hogy egy önszabályozó tanuló rendszer konvergál az optimálishoz, ha teljesen szabadon engedjük?”**

Ez valóban egy **alapkutatási értékű** kérdés. Vegyük ezt górcső alá az iménti struktúra szerint.

---

### 🔍 1. Mit jelent itt az "optimális"?

Ez lehet:

* **Minimális költségfüggvény** (pl. LQR-típusú kvadratikus költség)
* **Pontosság a rendszer becslésében** (kis hiba $\|A_k - A_{\text{true}}\|$)
* **Stabilitás + konvergencia egy célállapothoz**

Döntsük el, melyik aspektusra koncentrálunk. Elsőként talán a **becsült modell stabilitására** és **konvergenciájára** fókuszálhatunk, mert ez közvetlenül visszahat a szabályozásra is.

---

### 🔎 2. Mi akadályozza ezt a konvergenciát?

Ez egy kulcskérdés. Itt jön be a **"persistence of excitation"** elmélet:

* Ha az input $u[k]$ nem elég változatos, akkor az $X$ mátrixból nem lehet jó $A, B$-t becsülni.
* De ha az $u[k]$ **maga a rendszer kimenetére válasz** (zárt hurkú), akkor **nem garantált**, hogy az $X$ elég információt tartalmaz.

Ez azt jelenti, hogy egy teljesen "szabadon engedett" szabályozó **nem biztos, hogy begyűjti a szükséges információt a tanuláshoz**.

Ez a **visszacsatolt tanulás egyik fő paradoxona**:

> Jó szabályozás → stabil pálya → kevés variancia → rossz tanulás

---

$$

## 🔧 3. Egyszerűsített analitikus modell – az általad megadott rendszer alapján

Az alábbi modell absztrakcióját adom a rendszerednek, hogy *matematikailag vizsgálhatóvá váljon*:

---

### 🎯 **1. A rendszer dinamikája – absztrakció**

Az autó vertikális állapotát írjuk le egy (lokálisan) lineáris dinamikával:

$$
y[k+1] = y[k] + \Delta y[k] = y[k] + B u[k]
$$

Itt:

* $y[k] \in \mathbb{R}$: az aktuális pozíció az úton (a középtől mért eltérés),
* $u[k] \in [-10, 10]$: a vezérlő által kiválasztott lépés,
* $B$: ismeretlen, tanult dinamika – becsléssel határozzuk meg OLS-sel.

> Ez megfelel a szimulátorodban:
> `y_new = y + delta_y`,
> ahol a regresszió a `delta_y`-t becsüli, három szenzorból.

---

### 🧠 **2. A regresszió célja:**

A lineáris regresszió célja: becsülje a szenzoradatok és a delta\_y közötti kapcsolatot.
Ezt írhatjuk így:

$$
\Delta y[k] \approx \hat{B}^\top s[k]
$$

Ahol:

* $s[k] = [d_\text{left}, d_\text{center}, d_\text{right}]^\top$: szenzorbemenetek (előre néző),
* $\hat{B} \in \mathbb{R}^3$: regressziós súlyok (tanultak),
* A becslést minden időlépésben frissítjük:

  $$
  \hat{B}_k = \arg\min_{B} \sum_{i=0}^k \left( \Delta y[i] - B^\top s[i] \right)^2
  $$

Ez tehát egy **időben változó regressziós modell** zárt hurkú mintavételezéssel.

---

### 🧠 **3. A neurális háló célja:**

A második modell (MLP) célja:
a három szenzor aktuális becsült értéke alapján megjósolni, hogy milyen távolságban lesz az autó a középtől:

$$
\hat{z}[k] = \text{MLP}(s^\text{pred}[k])
$$

Itt:

* A MLP bemenete: $s^\text{pred}[k]$ = regresszióval becsült szenzoreltolódások
* A cél: $z[k] = y[k] - \text{center}[x[k]]$
* A tanítás: gradiens-alapú súlyfrissítés minősített példákon (scaled)

---

### 🔁 **4. A visszacsatolt tanulási dinamika:**

A vezérlő kiválasztja azt az $u[k]$-t, amely minimalizálja a várható eltérést a középtől:

$$
u[k] = \arg\min_{u} |\hat{z}[k+1]|
$$

Ez a szabályozási döntés:

* **függ a regressziós modell pontosságától** (a prediktált szenzorértékeken),
* **függ az MLP predikciójától**,
* és **meghatározza a következő állapotot**, tehát **minta-befolyásolt tanulás történik**.
