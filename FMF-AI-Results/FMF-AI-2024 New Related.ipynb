### Online Learning and Adaptive Sampling Strategies

In the context of adaptive sample selection beyond curriculum paradigms, **Bouchard et al. (2015)** introduce *Accelerating Stochastic Gradient Descent via Online Learning to Sample*, where they propose the **Adaptive Weighted SGD (AW‑SGD)**. This method jointly learns the model parameters and a sampling distribution that minimizes gradient variance on the fly. Their experiments on image classification, matrix factorization, and reinforcement learning tasks demonstrate **significantly faster convergence** compared to uniform sampling \cite{bouchard2015online} ([arXiv][1]). This result is directly relevant to our work: akin to AW‑SGD, in our setting the controller’s decision-making shapes sample generation, and minimizing gradient variance via structured perturbations can similarly stabilize learning.

More recently, **Zhao et al. (2025)** study **adaptive client sampling** in federated learning using an online bandit approach—optimizing which clients to train on to accelerate convergence in stochastic optimization. While in a federated context, their approach provides a conceptual analogue to closed‑loop supervised settings: sample selection is guided by feedback signals and uncertainty, a principle our VM‑allocation system naturally embodies \cite{zhao2025adaptive}.

In feedback-driven data generation, **Joulani et al. (2013)** investigate *Online Learning under Delayed Feedback*, presenting a meta‑algorithm that adapts standard online methods to uncertain delay in feedback. They derive regret bounds showing that in adversarial settings regret increases multiplicatively with delay, while in stochastic regimes only additively—delays thus scale overhead but remain controllable \cite{joulani2013delayed} ([arXiv][2]). This theory is directly relevant to our context, where control-dependent delays in observed outputs affect which samples are received when—understanding these effects is key to ensuring stability in closed‑loop supervised learning.

### ⚙️ Related Work (További kiegészítés)

In addition to existing works on curriculum learning and concept drift, a notable set of studies connects **Model Predictive Control (MPC)** to **online learning frameworks**, directly relevant to feedback-coupled supervised systems like ours.

**Wagener et al. (2019)** introduce the *Dynamic Mirror Descent Model Predictive Control (DMD‑MPC)* framework, which treats MPC as an instantiation of **online learning with dynamic regret**. The algorithm adapts the control policy iteratively using online updates based on dynamic mirror descent (a generalization of mirror descent suitable for evolving loss functions) and guarantees low regret over non-stationary cost sequences—demonstrated on both cart-pole simulations and real-world aggressive autonomous driving tasks on an AutoRally platform \cite{wagener2019dmdmpc} ([homes.cs.washington.edu][1]).

#### Elért eredmények:

* Megmutatták, hogy sok hagyományos MPC algoritmus **egyszerű speciális eset** DMD‑MPC-ben.
* Tiszta **online-learning perspektívát** kínálnak MPC-re, ahol minden lépés egy aktualizált döntési vektor meghatározása a következő round veszteségfüggvénye alapján.
* Kísérletileg igazolták, hogy az ilyen adaptív online beállítás **javítja a vezérlő teljesítményét és stabilitását**, különösen nemlineáris vagy változó környezetben.

#### Kapcsolódás a te kutatásodhoz:

* A te rendszereidben is a **vezérlő döntése alakítja a következő minta dağılitását**, hasonlóan az MPC-szcenárióhoz.
* A DMD‑MPC szemlélet és a dynamic regret formalizálás szoros párhuzamba állítható azzal, amikor a **vezérlési perturbációk csökkentik a mintavételi torzítást**, és ezáltal szabályozhatóbb konvergenciát eredményez.

## 🔍 Áttekintés: Control-Theoretic Techniques for Online Adaptation of Deep Neural Networks in Dynamical Systems

### 1. 📚 Háttér és motiváció

* A szerzők felvetése, hogy a szokásos DNN-eket (mély neurális hálókat) offline tanítják, de amikor valós környezetben használják őket, gyakran **domain shift** (visszaesés a teljesítményben) alakul ki.
* A megoldásuk: vezessünk be **kontroll-elméleti módszereket**, amelyek garantálják a DNN stabil és megbízható működését **online**, futás közben, a rendszer viselkedését követve ([Google Scholar][1], [arXiv][2], [arXiv][3]).

---

### 2. 🎯 Technikai koncepció

* A teljes hálót **folyamatos időben működő dinamikus rendszerként** kezelik.
* Csak a **legvégső (output) réteget** finomhangolják online, míg a korábbi rétegek maradnak statikusak (előzetesen offline betanítottak) ([arXiv][3], [Moonlight][4]).

---

### 3. ⚙️ Frissítési szabályok: Super-Twisting algoritmus

* A frissítési szabály a **“super-twisting algorithm”** (STA), amely egy egyszerű, folytonos, sliding-mode kontroll szabály, bizonyíthatóan **gyors hibacsökkentést** eredményez még zajos vagy változó környezetben is ([arXiv][3]).
* Két esetet tárgyalnak:

  * **Ismert** bemenet derivált mellett – teljes konvergencia (0 hibához).
  * **Becslés vagy zajos derivált esetén** – hibák *ultimately bounded*, azaz egy végső maximális téreltérés mellett stabilan tartják a rendszert ([arXiv][3], [Moonlight][4]).

---

### 4. 🛠️ Spectral Normalization (SN) garanciák

* A tanulás során alkalmazott **spektrális normalizáció** biztosítja, hogy a DNN **Lipschitz konstansát** felső korlátok között tartsa.
* Ezáltal **zajos input-ok vagy becsült deriváltak** esetén is kiszámítható, felső határ alatti hibák tarthatók ([arXiv][3], [Moonlight][4]).

---

### 5. 🧪 Empirikus validáció: Van der Pol oszcillátor

* Szimulációk során a DNN-et 4 rétegben, divers architektúrával képezték egy **nominal rendszer** alapján, majd egy paraméterében megváltozott (“real”) rendszerben tesztelték.
* Az online frissítések hatékonyan korrigálták a domain-shift hatását, és **a SN használatával kisebb konvergencia-határt** is elértek ([arXiv][3]).

---

### 📌 Hogyan illik be a te munkádhoz?

* A cikk pontosan azt a esszenciális problémát célozza, amit te is vizsgálsz: **DNN-reak adaptív használata zárt hurkú rendszerben, domain-shift esetén, online frissítéssel**.
* A nálad alkalmazott **perturbációs stratégiák mintaválasztásra** analóg módon a cikk **kontroll-elméleti update szabályokat** alkalmaz súlyfrissítésre.
* Fókusz: **one-pass streaming tanulás**, nincs újratanítás, de mégis biztosítanak stabil adaptációt—csakúgy, mint te a VM-allokáció kontroll révén.

**Javasolt bővített változat:**

Much prior work in supervised learning has investigated the effect of sample selection and ordering on convergence. Curriculum learning, hard negative mining, and stratified sampling methods are some well-known examples for these. But these typically rely on predefined sampling strategies to improve convergence speed or model robustness \cite{alain2016understanding, zhao2014stratified}.
Bouchard et al. introduce an online approach to dynamically select training examples based on the current model state, improving learning efficiency in evolving data contexts \cite{bouchard2015online}, which is conceptually related to our closed-loop sample selection.

Zhao and Zhang demonstrated that minibatch stratified sampling—partitioning data into clusters and sampling accordingly—can significantly reduce variance and accelerate convergence in SGD \cite{zhao2014stratified}. Alain and Bengio studied layer-wise linear classifier probes to understand intermediate representations, shedding light on how model internals impact learning dynamics \cite{alain2016understanding}.

The curriculum learning paradigm itself has been more deeply explored in recent work. Hacohen and Weinshall (2019) systematically examine how scoring and pacing strategies affect deep network convergence, showing substantial gains in both accuracy and training speed via curriculum schedules \cite{hacohen2019curriculum}. Graves et al. (2017) propose an automated curriculum learning framework where a multi-armed bandit selects training samples to maximize learning progress, significantly accelerating learning in sequence modeling tasks \cite{graves2017automated}.
Liu et al. present CL4CO, a curriculum-based training framework for graph neural combinatorial optimization that actively controls the sample order to improve convergence \cite{li2023selective}, which parallels our perturbation-based approach to diversify training trajectories.

General surveys such as Soviany et al. (2021) provide a taxonomy of curriculum learning methods, including self-paced and RL-guided curricula, while Narvekar et al. (2020) frame curriculum generation in reinforcement learning domains as a formal, adaptive process suitable for sequential decision-making settings. These works support the view that careful ordering of training examples can act as a powerful regularizer in dynamic systems.

In online learning and streaming data settings, concept drift adaptation has been extensively studied. Gama et al. presented a comprehensive survey on concept drift adaptation methodologies, including ensemble and drift detection techniques \cite{gama2014survey}. Elwell and Polikar proposed incremental ensemble methods for nonstationary environments that dynamically adapt to drift \cite{elwell2011learnpp}. Wang et al. explored mining ensemble classifiers for evolving data streams in KDD-2003 \cite{wang2003awe}. Losing et al. introduced a dual-memory architecture to manage diverse drift types in real-time systems \cite{losing2017selfadjusting}.
Meister and Nietert explore online estimation of sample statistics from comparison feedback \cite{meister2021comparison}, which is relevant when only partial performance information is available in feedback-coupled learning. Joulani et al. analyze online learning under delayed feedback \cite{joulani2013delayed}, a setting akin to control loops where the true outcome of actions becomes available only after a time lag.

Handling unexpected or out-of-distribution inputs is another critical aspect in such systems; a recent survey \cite{oodsurvey2025} summarizes strategies for robust generalization under distributional shift. Zhao et al. propose an adaptive client sampling method for federated learning using online bandit feedback to optimize participant selection \cite{zhao2025adaptive}, illustrating the value of dynamic sampling policies under evolving data distributions. Similarly, Xie et al. address semi-supervised one-pass learning under distribution shift \cite{cao2024semionline}, directly relating to our focus on single-pass adaptation in nonstationary environments.

While reinforcement learning naturally addresses closed-loop feedback dynamics, fewer studies consider purely supervised settings where the training data distribution is influenced by the model's own predictions. Our work focuses on these feedback-coupled systems without relying on reward signals or repeated training episodes. It aligns with adaptive control and online learning under concept drift, but prioritizes one-pass supervised updates in continuous feedback systems.

# Két rejtett rétegű MLP (10 és 5 neuron) – „kapacitás” számolása

## 1) Paraméterszám (W)

Jelölések: bemeneti dimenzió $d$, rejtett rétegek: $h_1=10$, $h_2=5$, kimenet $m$ (pl. regressziónál $m=1$).

Teljes tanulható paraméterszám:

$$
W \;=\; d\cdot h_1 + h_1 \;+\; h_1\cdot h_2 + h_2 \;+\; h_2\cdot m + m.
$$

Behelyettesítve $h_1=10, h_2=5$:

$$
W \;=\; 10d + 10 \;+\; 50 + 5 \;+\; 5m + m \;=\; 10d + 65 + 6m.
$$

Példa $m=1$ mellett: $W = 10d + 71$.

További hasznos mennyiségek:

* Rétegszám (súlyrétegek): $L=3$ (két rejtett + kimenet).
* Nemlineáris egységek száma: $U = h_1 + h_2 = 15$ (a kimenet általában lineáris).

## 2) VC-dimenzió / pseudodimenzió – mi számít az aktivációnál?

Igen, **számít az aktiváció típusa**. A legrelevánsabb, széles körben hivatkozott eredmények:

* **Darabolt lineáris (pl. ReLU)** hálóknál szűk (majdnem pontos) felső és alsó korlátok ismertek:

  $$
  \mathrm{VCdim} \;=\; O\!\big(W\,L\,\log W\big), \quad
  \mathrm{VCdim} \;=\; \Omega\!\big(W\,L\,\log(W/L)\big),
  $$

  és ekvivalensen (más parametrizálással) egy **szoros** korlát:

  $$
  \mathrm{VCdim} \;=\; \Theta(W\,U).
  $$

  Ezek **pseudodimenzióra** (valós kimenet, azaz regresszió) is átvihetők. ([Proceedings of Machine Learning Research][1], [jmlr.org][2])

* **Sima szigmoid/tanh** esetén régebbi általános korlátok azt mutatják, hogy a VC-dimenzió **polinomiálisan korlátos** a paraméterszámban (klasszikus $O(W^2)$ típusú felső korlátok általános osztályokra), és léteznek explicit polinomiális korlátok szigmoid/Pfaff-féle aktivációkra. Ezek gyengébbek (lazábbak), mint a ReLU-ra kapott majdnem szoros becslések. ([SpringerLink][3], [ScienceDirect][4], [theory.cs.uni-bonn.de][5])

**Mit jelent ez a te MLP-dre (ReLU-val):**

* Használhatod a két formát is:

  * Mélységfüggő: $\mathrm{VCdim} = O(W\,L\,\log W)$.
  * „Neuron-szám” függő: $\mathrm{VCdim} = \Theta(W\,U)$.

Mindkettő ugyanarra a nagyságrendre vezet; a második (Θ(W·U)) különösen kényelmes, mert itt $U=15$.

## 3) Gyors nagyságrendi becslések (ReLU-val)

Tegyük fel $m=1$.

* **Általános képlet**:

  * $W = 10d + 71$, $L=3$, $U=15$.
  * $\mathrm{VCdim} \approx \text{nagyságrendben } W\,L\,\log W$ **vagy** $\approx W\,U$ (konstans szorzók rejtve).

* **Számok például**:

  * $d=20$: $W=271$ → $W\,L\,\log W \approx 4.6\times10^3$, $W\,U = 4065$.
  * $d=100$: $W=1071$ → $W\,L\,\log W \approx 2.24\times10^4$, $W\,U = 1.61\times10^4$.

Ezek **nem éles határok**, hanem elméleti **nagyságrendi** kapacitásmutatók. A regressziós **pseudodimenzió** is ugyanezen nagyságrendbe esik ReLU esetén. ([jmlr.org][2])

Jó gyakorlat a következőkre hivatkozni (ezek ma a „standard” referenciák):

* **ReLU / darabolt lineáris aktivációk – szinte éles korlátok:**

  * Harvey, Liaw, Mehrabian (COLT 2017): $O(WL\log W)$ felső és $\Omega(WL\log(W/L))$ alsó korlát; darabolt lineáris aktivációkra, és kiterjeszthető a pseudodimenzióra. ([Proceedings of Machine Learning Research][1])
  * Bartlett, Harvey, Liaw, Mehrabian (JMLR 2019): részletes, journal-verzió; több megfogalmazás, köztük a $\Theta(WU)$ jellegű korlát is darabolt polinomokra. ([jmlr.org][6])

Két rejtett réteg esetén, $h_1, h_2$ rejtett neuronnal:

$$

Lépésenként:

* $d h_1 = 3\cdot 10 = 30$
* $+h_1 = +10 \Rightarrow 40$
* $+h_1 h_2 = +10\cdot 5 = +50 \Rightarrow 90$
* $+h_2 = +5 \Rightarrow 95$
* $+h_2 m = +5\cdot 1 = +5 \Rightarrow 100$
* $+m = +1 \Rightarrow \mathbf{W=101}$

tanh hálókra a klasszikus VC-dim/pseudodim felső korlát **polinomiális** a paraméterszámban; széles körben használt nagyságrendi mérő a **$W^2$**-skála.
$\Rightarrow$ „kvadratikus kapacitás-mutató”: $\;W^2 = 101^2 = \mathbf{10\,201}$.

#### 2) MLP (5, 3), $d=3, m=1$

* $d h_1 = 3\cdot 5 = 15$
* $+h_1 = +5 \Rightarrow 20$
* $+h_1 h_2 = +5\cdot 3 = +15 \Rightarrow 35$
* $+h_2 = +3 \Rightarrow 38$
* $+h_2 m = +3\cdot 1 = +3 \Rightarrow 41$
* $+m = +1 \Rightarrow \mathbf{W=42}$

$\Rightarrow$ $W^2 = 42^2 = \mathbf{1\,764}$.

**Beilleszthető szöveg (rövid, technikai):**

A kétrétegű MLP paraméterszáma $W = d h_1 + h_1 + h_1 h_2 + h_2 + h_2 m + m$. A nálunk használt két architektúra esetén $d=3, m=1$: (10,5) → $W=101$, (5,3) → $W=42$. A tanh aktivációjú hálók VC-/pseudodimenziójára klasszikus, széles körben hivatkozott, polinomiális (kvadratikus) felső korlát ismert $\tilde{O}(W^2)$ nagyságrendben; ezt itt csak nagyságrendi mutatóként használjuk (101² ≈ 10 201, 42² ≈ 1 764), a relatív kapacitás szemléltetésére.
A kapacitás önmagában nem garantál jobb online általánosítást; a mi zárt hurkú beállításunkban a minták eloszlása a vezérlő döntései miatt összehúzódik, ami torzíthatja a tanulási pályát (ld. "distributional narrowing").

Miért hasznos: olvasó azonnal látja, hogy a (10,5) háló **kb. 6×** nagyobb "kvadratikus" kapacitású, mégis instabilabb lehet autonom módban — ez jól rámutat arra, hogy **closed-loop** környezetben a kapacitás vs. mintázat lefedés dialektikája döntő.

---

### 2) "Closed-loop covariate shift" elnevezés (Terminológia tisztítás)

**Hová:** 1. és 3. fejezet végére, ahol az "distributional narrowing" fogalom megjelenik.

**Beilleszthető szöveg:**

A "distributional narrowing" jelenséget a tágabb irodalomban tekinthetjük a closed-loop covariate shift egy speciális esetének: a predikció → akció → állapotlánc endogén módon alakítja a tanulási eloszlást, és a lefedett állapottér fokozatosan beszűkül. Ezt a munkában kísérletileg dokumentáljuk és egyszerű, periodikus "dither"-perturbációval enyhítjük.

Miért hasznos: összeköti a saját fogalmat egy ismertebb ernyőfogalommal, megkönnyítve a hivatkozást és a befogadást.

---

### 3) **Állapottér-lefedettség mérőszám** (egy objektív coverage-metrika)

**Hová:** 6–7. fejezet (Results) elejére, a hibametrikák mellé.

**Beilleszthető szöveg:**

A tanulási pálya lefedettségét a $\{x[k]\}$ sorozat kovarianciamátrixának log-det értékével (elliptikus térfogat), illetve rácsalapú occupancy-aránnyal (hány rácscella látogatott) kvantifikáljuk. A periodikus perturbációtól **nő** a lefedettség (↑log-det, ↑occupancy), és ezzel csökken a futások közötti hibaszórás — összhangban a 2×2 kísérlet variancia-eredményeivel.

Miért hasznos: számszerűsíti azt, amit a görbék szemre mutatnak (stabilitásjavulás és kisebb szórás perturbációval).
